### Cleaning Users Data and Converting it into CSV

```python
import pandas as pd
import re
import numpy as np
import json

# Load data
file_path = '../Data_Code_Data_Folder/users.json'
with open(file_path, "r") as f:
    users_data = json.load(f)
users_df = pd.DataFrame(users_data)

# --- Cleaning 'name' column ---
# Drop rows where 'user_id' or 'name' are null (initial drop based on previous steps)
users_df = users_df.dropna(subset=['user_id', 'name'])
# Remove titles like Dr., Mrs., Mr., Ms. from the 'name' column
users_df['name'] = users_df['name'].astype(str).str.replace(r'(Dr\.|Mrs\.|Mr\.|Ms\.)\s*', '', regex=True)
# Drop rows where 'name' is an empty string
users_df = users_df[users_df['name'] != '']

# --- Cleaning 'age' column ---
# Calculate the average age, excluding nulls
average_age = users_df['age'].mean()
# Fill null values in the 'age' column with the calculated average age
users_df['age'].fillna(average_age, inplace=True)
# Round the age column and convert to integer
users_df['age'] = users_df['age'].round().astype(int)

# --- Cleaning 'email' column ---
# Replace 'invalid_email' with 'Not Provided'
users_df['email'] = users_df['email'].replace('invalid_email', 'Not Provided')
# Handle potential null values after replacement (though fillna might be more robust)
users_df['email'].fillna('Not Provided', inplace=True)
# Handle potential empty strings after replacement
users_df['email'].replace('', 'Not Provided', inplace=True)


# --- Cleaning 'address' column ---
# Replace null values in 'address' column with 'Not Provided'
users_df['address'].fillna('Not Provided', inplace=True)
# Replace empty strings in 'address' column with 'Not Provided'
users_df['address'].replace('', 'Not Provided', inplace=True)

# --- Cleaning 'device_type' column ---
# Replace null values and empty strings with 'Unknown'
users_df['device_type'].fillna('Unknown', inplace=True)
users_df['device_type'].replace('', 'Unknown', inplace=True)
# Standardize the case (capitalize the first letter)
users_df['device_type'] = users_df['device_type'].str.capitalize()

# --- Cleaning 'listening_time' column ---
# Coerce errors to turn non-numeric values into NaN
users_df['listening_time'] = pd.to_numeric(users_df['listening_time'], errors='coerce')
# Calculate the average listening time, excluding nulls
average_listening_time = users_df['listening_time'].mean()
# Fill null values with the calculated average listening time
users_df['listening_time'].fillna(average_listening_time, inplace=True)
# Round the listening_time column and convert to integer
users_df['listening_time'] = users_df['listening_time'].round().astype(int)

# --- Cleaning 'subscription_type' column ---
# Standardize the case of 'subscription_type'
users_df['subscription_type'] = users_df['subscription_type'].astype(str).str.capitalize()
# Drop rows with null or empty strings in 'subscription_type'
users_df.dropna(subset=['subscription_type'], inplace=True)
users_df = users_df[users_df['subscription_type'] != '']


print("✅ All specified cleaning steps applied to users_df.")
display(users_df.head())
display(users_df.info())
```

```python
# Save cleaned users_df to a CSV file
users_df.to_csv('users.csv', index=False)
print("✅ Cleaned users_df saved to 'users.csv'")
```

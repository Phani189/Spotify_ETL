### Cleaning Artists Data and Converting into CSV

```python
import pandas as pd
import json
import numpy as np
# Load data
file_path=file_path = '../Data_Code_Data_Folder/artists.json'
with open(file_path, "r") as f:
    artists_data = json.load(f)
artists_df = pd.DataFrame(artists_data)

# Drop rows where 'name' is null or an empty string
artists_df.dropna(subset=['name'], inplace=True)
artists_df = artists_df[artists_df['name'] != '']

# Handle null values in other object columns if any (after dropping rows with null names)
artists_df['country'] = artists_df['country'].fillna('Not Specified')
artists_df['genre'] = artists_df['genre'].fillna('No Genre')


# Standardize the case of the 'genre' column
artists_df['genre'] = artists_df['genre'].astype(str).str.capitalize()

# Handle null values and convert 'popularity' to numeric
artists_df['popularity'] = pd.to_numeric(artists_df['popularity'], errors='coerce')
average_popularity = artists_df['popularity'].mean()
artists_df['popularity'].fillna(average_popularity, inplace=True)

# Convert the 'popularity' column to integer type
artists_df['popularity'] = artists_df['popularity'].round().astype(int)


print("✅ Cleaning steps applied to artists_df.")
display(artists_df.head())
display(artists_df.info())
```

```python
# Save cleaned artists_df to a CSV file
artists_df.to_csv('artists.csv', index=False)
print("✅ Cleaned artists_df saved to 'artists.csv'")
```

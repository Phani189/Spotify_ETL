### Cleaning Songs Data and convert it into CSV

```python
import pandas as pd
import numpy as np
import json

# Load data
file_path='../Data_Code_Data_Folder/songs.json'
with open(file_path, "r") as f:
    songs_data = json.load(f)
songs_df = pd.DataFrame(songs_data)

# Drop rows where 'name' is null or empty string
songs_df.dropna(subset=['name','song_id'], inplace=True)
songs_df = songs_df[songs_df['name'] != '']

# Drop rows where 'album' is null or empty string
songs_df.dropna(subset=['album'], inplace=True)
songs_df = songs_df[songs_df['album'] != '']

# Handle null values and empty strings in 'genre' column (keeping fill for genre)
songs_df['genre'] = songs_df['genre'].fillna('No Genre').replace('', 'No Genre')

# Standardize the case of the 'genre' column
songs_df['genre'] = songs_df['genre'].astype(str).str.capitalize()

# Handle null values and the string 'short' in the 'duration' column
songs_df['duration'] = songs_df['duration'].replace('short', np.nan)
songs_df['duration'] = pd.to_numeric(songs_df['duration'], errors='coerce')
average_duration = songs_df['duration'].mean()
songs_df['duration'].fillna(average_duration, inplace=True)

# Convert the 'duration' column to integer type
songs_df['duration'] = songs_df['duration'].round().astype(int)

print("✅ Cleaning steps applied to songs_df.")
display(songs_df.head())
display(songs_df.info())
```

```python
# Save cleaned songs_df to a CSV file
songs_df.to_csv('songs.csv', index=False)
print("✅ Cleaned songs_df saved to 'songs.csv'")
```

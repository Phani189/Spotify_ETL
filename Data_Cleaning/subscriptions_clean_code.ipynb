### Subscriptions table cleaning and Convert it into CSV


```python
import pandas as pd
import numpy as np
import json

# Load data
file_path = '../Data_Code_Data_Folder/subscriptions.json'
with open(file_path, "r") as f:
    subscriptions_data = json.load(f)
subscriptions_df = pd.DataFrame(subscriptions_data)

# Handle null values and standardize case for 'subscription_type' and 'subscription_category'
subscriptions_df['subscription_type'] = subscriptions_df['subscription_type'].fillna('Free Trail').str.capitalize()
subscriptions_df['subscription_category'] = subscriptions_df['subscription_category'].fillna('No Subscription').str.capitalize()

# Handle null values and convert 'amount' to numeric
subscriptions_df['amount'] = pd.to_numeric(subscriptions_df['amount'], errors='coerce')
average_amount = subscriptions_df['amount'].mean()
subscriptions_df['amount'].fillna(average_amount, inplace=True)

# Set 'amount' to 0 for 'Free' or 'Free Trail' subscription types
subscriptions_df['amount'] = np.where(
    subscriptions_df['subscription_type'].isin(['Free', 'Free Trail']),
    0,
    subscriptions_df['amount']
)
subscriptions_df['amount'] = subscriptions_df['amount'].round(2)

# Convert date columns to datetime objects
subscriptions_df['start_date'] = pd.to_datetime(subscriptions_df['start_date'], errors='coerce')
subscriptions_df['end_date'] = pd.to_datetime(subscriptions_df['end_date'], errors='coerce')

print("✅ Cleaning steps applied to subscriptions_df.")
display(subscriptions_df.head())
display(subscriptions_df.info())
```

```python
# Save cleaned subscriptions_df to a CSV file
subscriptions_df.to_csv('subscriptions.csv', index=False)
print("✅ Cleaned subscriptions_df saved to 'subscriptions.csv'")
```

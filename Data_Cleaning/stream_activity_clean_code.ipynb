### Cleaning Stream_Activity and Convert it into CSV

```python
import pandas as pd
import numpy as np
import json

file_path='../Data_Code_Data_Folder/stream_activity.json'

# Load data
with open(file_path, "r") as f:
    stream_activity_data = json.load(f)
stream_activity_df = pd.DataFrame(stream_activity_data)

# Handle null values and the string 'many' in the 'play_count' column
# First convert to numeric, coercing errors including 'many' to NaN
stream_activity_df['play_count'] = pd.to_numeric(stream_activity_df['play_count'], errors='coerce')
average_play_count = stream_activity_df['play_count'].mean()
# Now replace 'many' (which are now NaN) with the calculated average
# and fill any remaining NaNs (original nulls) or empty strings with 0
stream_activity_df['play_count'] = stream_activity_df['play_count'].fillna(average_play_count)
stream_activity_df['play_count'].replace('', 0, inplace=True) # Replace empty strings if any

stream_activity_df['play_count'] = stream_activity_df['play_count'].round().astype(int)


# Handle null values and the string 'varied' in the 'listening_time' column
stream_activity_df['listening_time'] = stream_activity_df['listening_time'].replace('varied', np.nan)
stream_activity_df['listening_time'] = pd.to_numeric(stream_activity_df['listening_time'], errors='coerce')
average_listening_time = stream_activity_df['listening_time'].mean()
stream_activity_df['listening_time'].fillna(average_listening_time, inplace=True)
stream_activity_df['listening_time'] = stream_activity_df['listening_time'].round().astype(int)

# Handle null values and empty strings in the 'device_type' column and standardize the case
stream_activity_df['device_type'] = stream_activity_df['device_type'].fillna('Unknown').replace('', 'Unknown').str.capitalize()

# Convert the 'timestamp' column to datetime objects
stream_activity_df['timestamp'] = pd.to_datetime(stream_activity_df['timestamp'])

# Handle null values in 'is_liked' and 'is_skipped' columns by replacing them with the mode
for col in ['is_liked', 'is_skipped']:
    mode_value = stream_activity_df[col].mode()[0]
    stream_activity_df[col].fillna(mode_value, inplace=True)

print("✅ Cleaning steps applied to stream_activity_df.")
display(stream_activity_df.head())
display(stream_activity_df.info())
```

```python
# Save cleaned stream_activity_df to a CSV file
stream_activity_df.to_csv('stream_activity.csv', index=False)
print("✅ Cleaned stream_activity_df saved to 'stream_activity.csv'")
```

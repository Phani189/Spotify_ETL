### Cleaning Albums Data and convert into CSV

```python
import pandas as pd
import json
import numpy as np

file_path=file_path = '../Data_Code_Data_Folder/albums.json'
# Load data
with open(file_path, "r") as f:
    albums_data = json.load(f)
albums_df = pd.DataFrame(albums_data)

# Drop rows where 'album_name' is null or empty string
albums_df.dropna(subset=['album_name','album_id'], inplace=True)
albums_df = albums_df[albums_df['album_name'] != '']

# Convert 'release_date' column to datetime objects
albums_df['release_date'] = pd.to_datetime(albums_df['release_date'])

print("✅ Cleaning steps applied to albums_df.")
display(albums_df.head())
display(albums_df.info())
```

```python
# Save cleaned albums_df to a CSV file
albums_df.to_csv('albums.csv', index=False)
print("✅ Cleaned albums_df saved to 'albums.csv'")
```

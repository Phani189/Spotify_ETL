### Data Generation code

```python
import json
import random
from faker import Faker
from datetime import datetime, timedelta

fake = Faker()

# Constants
NUM_USERS = 5000
NUM_ARTISTS = 40
NUM_SONGS = 800
NUM_FACTS = 100000
DEVICE_TYPES = ['Mobile', 'Desktop', 'Tablet', 'mobile', 'DESKTOP']
SUBSCRIPTION_TYPES = ['Free', 'Premium', 'Family', 'premium']
SUBSCRIPTION_CATEGORIES = ['Monthly', '3 Month', 'Yearly']
GENRES = ['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Classical', 'Electronic', 'Melody']

# Generate Users
users = []
for i in range(1, NUM_USERS + 1):
    users.append({
        "user_id": i,
        "name": fake.name() if random.random() > 0.001 else "",
        "age": random.choice([random.randint(18, 70), None]),
        "email": fake.email() if random.random() > 0.001 else "invalid_email",
        "address": fake.address().replace("\n", ", ") if random.random() > 0.001 else None,
        "device_type": random.choice(DEVICE_TYPES),
        "listening_time": random.choice([random.randint(100, 10000)]),
        "subscription_type": random.choice(SUBSCRIPTION_TYPES),
        "subscription_category": random.choice(SUBSCRIPTION_CATEGORIES)
    })

# Generate Subscriptions
subscriptions = []
for user in users:
    start_date = fake.date_between(start_date='-1y', end_date='today')
    category = user["subscription_category"]
    if category and category.lower() == "monthly":
        end_date = start_date + timedelta(days=30)
    elif category and category.lower() == "3 month":
        end_date = start_date + timedelta(days=90)
    elif category and category.lower() == "yearly":
        end_date = start_date + timedelta(days=365)
    else:
        end_date = start_date + timedelta(days=random.randint(30, 365))

    subscriptions.append({
        "user_id": user["user_id"],
        "subscription_type": user["subscription_type"],
        "subscription_category": category,
        "start_date": start_date.isoformat(),
        "end_date": end_date.isoformat(),
        "amount": random.choice([random.uniform(200,1000), "free", None])
    })

# Generate Artists
artists = []
for i in range(1, NUM_ARTISTS + 1):
    genre = random.choice(GENRES)
    artists.append({
        "artist_id": i,
        "name": fake.name() if random.random() > 0.001 else None,
        "country": fake.country(),
        "popularity": random.choice([random.randint(50, 100), None]),
        "genre": genre
    })

# Ensure every artist has at least one album
albums = []
album_id_counter = 1
for artist in artists:
    num_albums = random.randint(1, 5)
    for _ in range(num_albums):
        albums.append({
            "album_id": album_id_counter,
            "album_name": fake.word().capitalize() if random.random() > 0.001 else " ",
            "release_date": fake.date_between(start_date='-5y', end_date='today').isoformat(),
            "artist_id": artist["artist_id"]
        })
        album_id_counter += 1

# Generate Songs
songs = []
for i in range(1, NUM_SONGS + 1):
    artist = random.choice(artists)
    artist_albums = [a for a in albums if a["artist_id"] == artist["artist_id"]]
    if artist_albums:
        album = random.choice(artist_albums)
        songs.append({
            "song_id": i,
            "name": fake.sentence(nb_words=3) if random.random() > 0.001 else None,
            "album": album["album_name"],
            "album_id": album["album_id"],
            "artist_id": artist["artist_id"],
            "duration": random.choice([random.randint(5, 25), "short"]),
            "genre": artist["genre"]
        })

# Generate Fact Table
stream_activity = []
for _ in range(NUM_FACTS):
    user = random.choice(users)
    song = random.choice(songs)
    stream_activity.append({
        "user_id": user["user_id"],
        "artist_id": song["artist_id"],
        "song_id": song["song_id"],
        "play_count": random.choice([random.randint(1, 50), None, "many"]),
        "listening_time": random.choice([random.randint(10, 100), "varied"]),
        "device_type": user["device_type"],
        "album_id": song["album_id"],
        "timestamp": (datetime.now() - timedelta(days=random.randint(0, 365))).isoformat(),
        "is_liked": random.choice([True, False]),
        "is_skipped": random.choice([True, False])
    })

# Save to JSON files
with open("users.json", "w") as f: json.dump(users, f, indent=2)
with open("subscriptions.json", "w") as f: json.dump(subscriptions, f, indent=2)
with open("artists.json", "w") as f: json.dump(artists, f, indent=2)
with open("albums.json", "w") as f: json.dump(albums, f, indent=2)
with open("songs.json", "w") as f: json.dump(songs, f, indent=2)
with open("stream_activity.json", "w") as f: json.dump(stream_activity, f, indent=2)

# Summary logging
print("✅ JSON data files with uncleaned entries generated successfully.")
print(f"Users: {len(users)}")
print(f"Subscriptions: {len(subscriptions)}")
print(f"Artists: {len(artists)}")
print(f"Albums: {len(albums)}")
print(f"Songs: {len(songs)}")
print(f"Fact Table Records: {len(stream_activity)}")
```
